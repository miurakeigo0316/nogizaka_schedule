```python
def find_max(numbers):
    max_num=numbers[0]
    for num in numbers:
        if num > max_num:
            max_num =num
    return max_num

print(find_max([2,6,4,8,1]))
```

```python
def multiply_list(lis):
    result = 1
    for num in lis:
        result*=num
    return result

print(multiply_list([1,2,3,4,5]))
```
