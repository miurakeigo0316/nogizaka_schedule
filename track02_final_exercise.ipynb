```python
import pandas as pd
import japanize_matplotlib

df=pd.read_csv("t02_survey.csv")

df.dropna(inplace=True)

sales=df["部門"] =="営業"
df_sales=df[sales]

df_sales.groupby("月")["満足度"].mean().plot(kind="line",ylabel="満足度平均",title="営業部月ごと満足度平均値")
```
