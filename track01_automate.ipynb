```python
import os
import shutil

# ==============================
# 設定部分（ここを変更すれば使い回せます）
# ==============================

templete_folder = "会議_template"   # コピー元フォルダ
base_folder = "2025年度会議"        # 作成する親フォルダ名
start_year = 2025                  # 年度の開始年（2025年度なら2025）

# ==============================
# メイン処理
# ==============================

# 親フォルダを作成（すでにあればエラーにならない）
os.makedirs(base_folder, exist_ok=True)

# ------------------------------
# ① 4月〜12月（開始年）
# ------------------------------
for month in range(4, 13):  # 4〜12まで繰り返し
    folder_name = f"{start_year}年{month:02}月"
    
    # 作成先のパスを作る
    target_folder_path = os.path.join(base_folder, folder_name)

    # テンプレートフォルダをコピー
    shutil.copytree(templete_folder, target_folder_path)

# ------------------------------
# ② 1月〜3月（翌年）
# ------------------------------
next_year = start_year + 1  # 翌年を計算

for month in range(1, 4):  # 1〜3まで繰り返し
    folder_name = f"{next_year}年{month:02}月"
    
    # 作成先のパスを作る
    target_folder_path = os.path.join(base_folder, folder_name)

    # テンプレートフォルダをコピー
    shutil.copytree(templete_folder, target_folder_path)

print("年度フォルダの作成が完了しました")
```
