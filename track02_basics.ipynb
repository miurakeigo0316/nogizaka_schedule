```python
import pandas as pd

# CSVファイルを読み込む
# ファイル名は同じフォルダにある前提
df = pd.read_csv("sales_data_v2.csv")

# 「地域」ごとの購入数をカウントする
region_counts = df["地域"].value_counts()

# 棒グラフを表示する
# Pandasのplot機能を使用（内部でmatplotlibが使われるが、直接インポートはしない）
region_counts.plot(kind="bar", title="地域ごとの購入数")

# グラフを表示する
```
