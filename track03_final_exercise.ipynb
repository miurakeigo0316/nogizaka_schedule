```python
import requests
import csv
import pandas as pd
import japanize_matplotlib

url= "https://b1cmc2eyo3.execute-api.ap-northeast-1.amazonaws.com/customers"

response=requests.get(url)

data=response.json()

df=pd.DataFrame(data)

df.dropna(inplace=True)

df["年代"].value_counts().plot(
    kind="pie",
    autopct="%.1f%%",
    title="参加者の年代別割合",
    counterclock=False,
    startangle=90,
    ylabel=""
)
```
